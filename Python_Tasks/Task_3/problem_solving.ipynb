Q1. Find the Frequency of Each Word

Pseudocode:
1. Start
2. Take a sentence as input.
3. Split the sentence into words.
4. Create an empty dictionary.
5. Check each word.
6. If the word exists, increase its count by 1.
7. Otherwise, add the word with count 1.
8. Return the dictionary.
9. Stop

Q10. Find Two Numbers Whose Sum Equals a Target

Pseudocode:
1. Start
2. Take a list and target as input.
3. Check each element from left to right.

Q13. Rotate a List to the Right by K Positions

Pseudocode:
1. Start
2. Take a list and K as input.
3. If the list is empty, return an empty list.
4. Calculate K modulo the list length.
5. Move the last K elements to the beginning.
6. Place the remaining elements after them.
7. Return the rotated list.
8. Stop
