```python
s = input("Enter string: ")

s = s.replace(" ", "")
freq = {}
result = []

for i in s:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

for i in s:
    if freq[i] > 1 and i not in result:
        result.append(i)

print(result)
```

Convert sentence to lowercase
Split into words
Count each word
Loop through words
If count == 1
    Print word
    Break

```python
#18
s = input("Enter sentence: ")

words = s.lower().split()
freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

for word in words:
    if freq[word] == 1:
        print(word)
        break
else:
    print("No unique word")
```
